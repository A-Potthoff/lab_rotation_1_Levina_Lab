```python
import numpy as np
import matplotlib.pyplot as plt
from src.utils.plotting import PET6, PET8, PET10, OWN

x = np.linspace(0, 1, 200)
for name, pal in {"PET6": PET6, "PET8": PET8, "PET10": PET10, "OWN": OWN}.items():
    print(f"\n{name}")
    pal.show()  # swatches: normal vision + deutan, protan, tritan
    fig, ax = plt.subplots(figsize=(8, 2.5), layout="constrained")
    for k, (n, c) in enumerate(pal.names.items()):  # all colours, incl. OWN's reserved grey
        ax.plot(x, np.sin(2 * np.pi * (x + k / len(pal.names))) + 0.2 * k, color=c, lw=2, label=n)
    ax.set(title=f"{name} in use", xticks=[], yticks=[])
    ax.legend(fontsize=7, loc="upper left", bbox_to_anchor=(1, 1))
    plt.show()
```
