````python
import pathlib
import datetime

def initialize_environment(author_name="Andre Potthoff", license_type="MIT", project_name="scientific_project"):
    """Builds the repository structure and generates standard boilerplate files."""
    
    base_dir = pathlib.Path.cwd()
    current_year = datetime.date.today().year

    # 1. Define the directory tree
    directories = [
        "archive",          # The "Zoo" for deprecated code snippets
        "data/raw",         # raw, unprocessed data
        "data/processed",   # processed data
        "data/external",    # other data sources from publications, and databases
        "notebooks",        # for development and analysis
        "src/config",       # configuration files, eg for simulations, model parameters, ...
        "src/models",       # model specifications, eg neural networks, ODEs, ...
        "src/simulation",   # simulation logic, running the models
        "src/analysis",     # functions for analysis of models/data/...
        "src/utils",        # eg plotting scripts
        "docs/annotations", # for longer derivations, explanations, and literature references
        "reports/figures",  # for plots, figures
        "results/objects",  # For heavy computed outputs (pickles, h5py)
        "models/saved",     # For trained model weights
        "tests",            # For unit tests
        "logs"
    ]

    # 2. Define License Text
    if license_type.upper() == "MIT":
        license_text = (
            f"MIT License\n\n"
            f"Copyright (c) {current_year} {author_name}\n\n"
            "Permission is hereby granted, free of charge, to any person obtaining a copy "
            "of this software and associated documentation files (the \"Software\"), to deal "
            "in the Software without restriction, including without limitation the rights "
            "to use, copy, modify, merge, publish, distribute, sublicense, and/or sell "
            "copies of the Software, and to permit persons to whom the Software is "
            "furnished to do so, subject to the following conditions:\n\n"
            "The above copyright notice and this permission notice shall be included in all "
            "copies or substantial portions of the Software.\n\n"
            "THE SOFTWARE IS PROVIDED \"AS IS\", WITHOUT WARRANTY OF ANY KIND, EXPRESS OR "
            "IMPLIED, INCLUDING BUT NOT LIMITED TO THE WARRANTIES OF MERCHANTABILITY, "
            "FITNESS FOR A PARTICULAR PURPOSE AND NONINFRINGEMENT. IN NO EVENT SHALL THE "
            "AUTHORS OR COPYRIGHT HOLDERS BE LIABLE FOR ANY CLAIM, DAMAGES OR OTHER "
            "LIABILITY, WHETHER IN AN ACTION OF CONTRACT, TORT OR OTHERWISE, ARISING FROM, "
            "OUT OF OR IN CONNECTION WITH THE SOFTWARE OR THE USE OR OTHER DEALINGS IN THE "
            "SOFTWARE.\n"
        )
    else:
        license_text = f"Copyright (c) {current_year} {author_name}. All rights reserved.\n"

    # 3. Define standard files and their initial contents
    files_to_create = {
        "README.md": (
            f"# Scientific Project by {author_name}\n\n"
            "## Architecture\n\n"
            "### Core Execution\n"
            "- `src/`: Core logic, mathematical models, simulations and functions.\n"
            "- `notebooks/`: Exploratory data analysis, prototyping, and result visualization.\n"
            "- `archive/`: for deprecated scripts and redundant logic that might be useful later.\n\n"
            "### Data & Output\n"
            "- `data/`: Local storage for datasets.\n"
            "- `results/objects/`: Save computed objects (e.g., `.pkl`, `.npy`, `.h5`).\n"
            "- `models/saved/`: eg. model weights and states.\n"
            "- `reports/figures/`: plots and figures\n"
            "- `logs/`\n\n"
            "### Documentation\n"
            "- `docs/annotations/`: Markdown explanations linked to source code. Used for mathematical derivations and literature references.\n\n"
            "## Setup\n"
            "This project uses a Conda environment.\n"
            "```bash\n"
            "conda config --set channel_priority strict\n"
            "conda env create -f environment.yml\n"
            "conda activate neuro_env\n"
            "pip install -e .\n"
            "```\n"
        ),
        
        f"environment.yml": (
            f"name: {project_name}\n"
            "channels:\n"
            "  - pytorch\n"
            "  - conda-forge\n"
            "  - defaults\n"
            "dependencies:\n"
            "  - python=3.10\n"
            "  - numpy\n"
            "  - scipy\n"
            "  - pandas\n"
            "  - h5py\n"
            "  - matplotlib\n"
            "  - jupyterlab\n"
            "  - python-dotenv\n"
            "  - pytest\n"
            "  - black[jupyter]  # Formats code in Python scripts and Jupyter cells\n"
            "  - pytorch\n"
            "  - torchvision\n"
            "  - torchaudio\n"
            "  - cpuonly         # install CPU-only dependencies without CUDA bloat\n"
            "  - pip\n"
        ),

        "pyproject.toml": (
            "[build-system]\n"
            "requires = [\"setuptools>=61.0\"]\n"
            "build-backend = \"setuptools.build_meta\"\n\n"
            "[project]\n"
            f"name = \"{project_name}\"\n"
            "version = \"0.1.0\"\n"
            f"authors = [{{ name=\"{author_name}\" }}]\n"
            "description = \"Computational neuroscience project\"\n"
            "readme = \"README.md\"\n"
            "requires-python = \">=3.10\"\n\n"
            "[tool.setuptools.packages.find]\n"
            "where = [\".\"]\n"
            "include = [\"src*\"]\n\n"
            "[tool.black]\n"
            "line-length = 88\n"
            "target-version = ['py310']\n"
        ),

        ".gitattributes": (
            "# use jupyter-aware diffing tools for notebooks\n"
            "*.ipynb diff=jupyternotebook\n"
            "*.ipynb merge=jupyternotebook\n"
        ),

        "docs/annotations/TEMPLATE.md": (
            "# Scientific Annotation Template\n\n"
            "### [Concept/Equation Name]\n"
            "**Context:**\n"
            "Why is this implemented this way?\n\n"
            "**Derivation:**\n"
            "$$\n\n$$\n\n"
            "**Source:**\n"
            "- Source: [Citation]\n\n"
            "---\n"
            "### Back-Reference Selection (Delete what you don't use)\n\n"
            "**Option A: Native Path (Cleanest, but fragile to file movement)**\n"
            "Use when linking to a specific function name or static file.\n"
            "[Return to implementation](../../src/simulation/biophysics.py)\n\n"
            "**Option B: The URI Search Tag (Most robust against code moving lines)**\n"
            "Requires a matching comment in the code like: `# [REF-UNIQUE-TAG] docs/annotations/file.md`\n"
            "[Return to Code](vscode://search?query=REF-UNIQUE-TAG)\n\n"
            "**Option C: Comment Anchors Extension (For global dashboards)**\n"
            "Requires the VS Code 'Comment Anchors' extension. `# MATH: Equation 1`\n"
        ),

        "LICENSE": license_text,

        ".gitignore": (
            "# Built environments\n"
            ".env\n"
            ".venv/\n"
            "env/\n"
            "venv/\n\n"
            "# Python caches and outputs\n"
            "__pycache__/\n"
            "*.py[cod]\n"
            "*$py.class\n"
            ".pytest_cache/\n\n"
            "# Jupyter Notebooks\n"
            ".ipynb_checkpoints\n\n"
            "# OS generated files\n"
            ".DS_Store\n"
            "Thumbs.db\n\n"
        )
    }

    # 4. Execute directory creation
    print("Building directories and module structure...")
    for dir_path in directories:
        target_dir = base_dir / dir_path
        target_dir.mkdir(parents=True, exist_ok=True)
        
        # Create a .gitkeep file so empty folders still get tracked by Git initially
        (target_dir / ".gitkeep").touch()
        
        # Turn the src folders into proper Python modules so autoreload works
        if dir_path.startswith("src/"):
            (target_dir / "__init__.py").touch()
            
        print(f"  + {target_dir.relative_to(base_dir)}")
        
    # Create the root src __init__
    (base_dir / "src" / "__init__.py").touch()

    # 5. Execute file generation
    print("\nGenerating boilerplate files...")
    for file_path, content in files_to_create.items():
        target_file = base_dir / file_path
        if not target_file.exists():
            target_file.write_text(content, encoding="utf-8")
            print(f"  + Created {target_file.relative_to(base_dir)}")
        else:
            print(f"  - Skipped {target_file.relative_to(base_dir)} (Already exists)")

    print("\nEnvironment successfully initialized.")
    print("\nNext steps to activate the bulletproof environment:")
    print("1. conda config --set channel_priority strict")
    print("2. conda env create -f environment.yml")
    print(f"3. conda activate {project_name}")
    print("4. pip install -e .")

if __name__ == "__main__":
    initialize_environment(author_name="Andre Potthoff", license_type="MIT", project_name="Levina_rotation_1")
````
